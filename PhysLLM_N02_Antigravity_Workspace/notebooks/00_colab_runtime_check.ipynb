2. 확장이 보이지 않으면 이 노트북을 Google Colab 웹에서 열어 실행합니다.
3. Python/PyTorch 상태만 확인합니다. 이 노트북에서 설치된 PyTorch를 임의로 업그레이드하지 마세요.


```python
import os, sys, platform, importlib.util, json
from pathlib import Path

print("Python:", sys.version)
print("Platform:", platform.platform())
print("CWD:", Path.cwd())
print("COLAB_RELEASE_TAG:", os.environ.get("COLAB_RELEASE_TAG", "not detected"))
for package in ["pandas", "transformers", "torch", "pytest", "nbformat"]:
    print(f"{package} installed:", importlib.util.find_spec(package) is not None)
```

```python
from google.colab import drive
from pathlib import Path

drive.mount("/content/drive")

PROJECT = Path(
    "/content/drive/MyDrive/PhysLLM/PhysLLM_N02_Antigravity_Workspace"
)

print("프로젝트 폴더 존재:", PROJECT.exists())

if PROJECT.exists():
    print(*sorted(p.name for p in PROJECT.iterdir()), sep="\n")
```

```python
# Inspect the existing runtime; do not install/upgrade torch automatically.
try:
    import torch
    print("torch:", torch.__version__)
    print("CUDA available:", torch.cuda.is_available())
    if torch.cuda.is_available():
        print("GPU:", torch.cuda.get_device_name(0))
        print("GPU memory GiB:", round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2))
except Exception as exc:
    print("PyTorch check unavailable:", repr(exc))
```
